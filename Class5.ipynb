```python
#Creates a sequence of vectors with uniform on [0,1) distributed components.

class RandomVectorLCG:
    def __init__(self, length, dimension):
        self._random = np.random.default_rng(3141) #1_D random number gen x_k
        self._randomVecSeq = np.zeros((length, dimension)) #d_D sequence [sampleIndex][componentIndex]
        
        for i in range(length):
            for j in range(dimension):
                #Populate the vector y_{i, j} = x_{i*d+j}
               self._randomVecSeq[i][j] = self._random.random()

    def getSeq(self): #getter
        return self._randomVecSeq

    def getValue(self, index): #getter
        return self._randomVecSeq[index]

    def getValueForComponent(self, index, component): #getter
        return self._randomVecSeq[index][component]

    def getSeqForComponent(self, componentIndex):
        componentSeq = np.zeros(len(self._randomVecSeq))
        for i in range(len(self._randomVecSeq)):
            componentSeq[i] = self._randomVecSeq[i][componentIndex]
        return componentSeq
    
    def graph(self):
        x = self._randomVecSeq[:, 0]
        y = self._randomVecSeq[:, 1]
        plt.figure(figsize=(10, 6))
        plt.scatter(x, y, alpha = 0.7)
        plt.title("Plot of two dimensions random vectors")
        plt.xlabel("x_1")
        plt.ylabel("x_2")
        plt.legend()
        plt.grid(True)
        plt.show()
```

```python
proof = RandomVectorLCG(100, 2)
proof.getSeq()
```

```python
proof.getValue(1)
```

```python
proof.getValueForComponent(1, 1)
```

```python
proof.getSeqForComponent(1)
```

```python
proof.graph()
```

# HOW A SEQUENCE OF IID RANDOM VARIABLES IS MODELED AS A PRODUCT SPACE
Given a random variable:

$$ X: \Omega \to \mathbb{R} $$

- $\Omega$ is the sample space (the set of all possible outcomes).
- $\mathcal{F}$ is the sigma-algebra (the set of events).
- $P$ is the probability measure (the function that assigns probabilities to events).

Suppose we want to obtain ( n )(size of the sampple from X ) independent realizations of ( X ). To model this formally, we construct the product space:

$$ \tilde{\Omega} := \Omega \times \cdots \times \Omega = \Omega^n $$

Each element of this space has the form:

$$ \tilde{\omega} = (\omega_1, \omega_2, \dots, \omega_n) \quad \text{with} \quad \omega_i \in \Omega $$

We define the product sigma-algebra:

$$ \tilde{\mathcal{F}} := \sigma(\mathcal{F} \times \cdots \times \mathcal{F}) $$

And the product probability measure:

$$ \tilde{P}\left( \prod_{i=1}^n A_i \right) = \prod_{i=1}^n P(A_i), \quad \forall A_i \in \mathcal{F} $$

- A is an event in the product space $( \tilde{\mathcal{F}} )$.

Next, we define a sequence of random variables:

$$ \tilde{X} = (\tilde{X}_1, \dots, \tilde{X}_n), \quad \text{where} \quad \tilde{X}_i(\tilde{\omega}) := X(\omega_i) $$

Thus, a single realization ( \tilde{\omega} \in \tilde{\Omega} ) produces an i.i.d. sample:

$$ (X(\omega_1), \dots, X(\omega_n)) $$

> In summary: To model multiple independent draws of a random variable ( X ), we use a single realization in the product space $( \tilde{\Omega} )$. This formalism is fundamental in simulation methods such as Monte Carlo, where independent samples are required.

> `Remember:` 
> - In our modeling doesn't care if we have 100 hundred of IID RV in 10 dimension space. (MC breaks the curse of dimensionality).
> - Many numerical methods have an overhead that scales exponentially in the dimension, but in MC we need *d* one-dimensional sample points to generate one d-dimensional sample point.
