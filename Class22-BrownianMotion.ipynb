# Recapitulation: Brownian Motion

**Definition 66 (Brownian Motion, family of random variables parametrized by time t):**  
Let $W : [0, \infty) \times \Omega \to \mathbb{R}^n$ denote a stochastic process with the following properties:

1. $W(0) = 0$ ($P$-almost surely).

2. The map $t \mapsto W(t)$ is continuous ($P$-almost surely$).

3. For given $t_0 < t_1 < \cdots < t_k$, the increments $W(t_1) - W(t_0), \ldots, W(t_k) - W(t_{k-1})$ are mutually independent.

4. For all $0 \leq s \leq t$ we have  
$$
W(t) - W(s) \sim \mathcal{N}(0, (t-s) I_n),
$$  
i.e., the increment is normally distributed with mean $0$ and covariance matrix $(t-s) I_n$ (*so, the variance gross linear in time*), where $I_n$ denotes the $n \times n$ identity matrix.

---
> For independent normal dist. r.v.: The variance of the sum is the sum of the variances.

> We can construct a numerical method of the value Brownian motion at a later time Tk: $W(t_k)=W(t_0) + \sum^{k-1}_{j=0} W(t_{j+1}) - W(t_{j})$

<p align="center">
<img src = 'image-34.png' width = '400' height = '200'>
</p>

Then $W$ is called an **$n$-dimensional $P$-Brownian motion** or an **$n$-dimensional $P$-Wiener process**.

>Only the first increment (from $T_0$ to $T_1$) is centered at $0$, because it starts from $W(T_0) = 0$ (by definition of Brownian motion).

> After that, each increment still follows a normal distribution with mean $0$, but the distribution of the total value $W(T_i)$ is no longer centered at $0$, it is centered around the value reached at the previous step.
