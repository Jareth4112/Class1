```python
tiny = 1.0
while tiny/2.0 > 0.0 and tiny/2.0 < tiny:
    tiny = tiny/2.0

print("tiny, ulp for 0", tiny)
print("tiny, ulp for 0", tiny/2)

print("\n","FLOATING POINT ARITHMETIC IS NOT COMMUTATIVE")
print("tiny/2 * 2:", (tiny/2)*2)
print("tiny*2 / 2:", (tiny*2)/2)
```

```python
eps = 1.0
prev_eps = eps  # Variable para almacenar el valor anterior de eps
while (1 + eps) > 1.0:
    prev_eps = eps  # Guarda el valor actual de eps
    eps = eps / 2.0

print("eps, ulp for 1", prev_eps)
print("eps, ulp for 1", 1+prev_eps)
```

```python
eps = 1.0
while (1 + eps > 1 and 1 + eps < 1 + 2*eps):
    eps = eps/2.0

print("eps", eps)
print("1 + eps", eps)
print("0.5*ulp(1)..", (0.5*np.spacing(1.0)))
print("0.5*(2^-52)..", (0.5*(2**-52))) 
```
