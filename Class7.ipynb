# **MONTE CARLO INTEGRATION:**
=============================================
- Let $f: \mathbb{R} \to \mathbb{R}$. If Xi are i.i.d. then Zi = f(Xi) are i.i.d.

- Example:
    - Consider a sequence of i.i.d. random variables {Xi} where $i \in \mathbb{N}$ having a uniform distribution on [0, 1]. Then, {f($X_i$)} where  $i \in \mathbb{N}$  is a sequence of i.i.d. random variables and the law of large numbers implies.

$$
\lim_{n \to \infty} \frac{1}{n} \sum_{i=1}^{n} f(X_i) \to E^p[f(X)] = \int_0^1 f(x) dx \space \text{almost surely}
$$

- So, first we need to see how other integration rules lies in the curse of the dimensionality. 


### **Simpson's Rule(Composite):**
- A classical approximation:
$$
\int_a^b f(x) \, dx \approx \tilde{I}_{\text{Smps}}(f; a, b) := \frac{b - a}{6} \left( f(a) + 4f\left(\frac{a + b}{2}\right) + f(b) \right)
$$

For the error of the Simpson’s rule we have:

$$
\left| \tilde{I}_{\text{Smps}}(f; a, b) - \int_a^b f(x) \, dx \right| \leq \frac{1}{90} \left( \frac{b - a}{2} \right)^5 \left| f^{(4)}(\xi) \right|,
$$


for some $\xi \in (a, b)$.

Discretizing and interval [c, d] into a m-intervals and applying the above rule to each subinterval, we get the composite Simpson's rule for $\int_{d}^{c} f(x) dx$

**USING m + 1 valuation points**

Using $(m+1)$ valuation points for the interval $[c, d]$, we obtain the composite Simpson’s rule:

$$
\int_c^d f(x) \, dx \approx \tilde{I}_{\text{Smps}}(f; c, d, m+1) := \frac{h}{3} \left( f(x_0) + 4 \sum_{j=1}^{m/2} f(x_{2j-1}) + 2 \sum_{j=1}^{m/2 - 1} f(x_{2j}) + f(x_m) \right),
$$

where $x_j = c + j\left(\frac{d - c}{m}\right)$.

For the error of the composite Simpson’s rule we have:

$$
\left| \tilde{I}_{\text{Smps}}(f; c, d, m+1) - \int_c^d f(x) \, dx \right| \leq \frac{h^4}{180}(d - c) \max_{\xi \in [c, d]} \left| f^{(4)}(\xi) \right|,
$$

where $( h := \frac{d - c}{m})$.

The formula uses $( m + 1 )$ evaluation points (m) intervals of size (h) and m+1 has to be odd.

### **Simpson's Rule in Higher Dimensions**

We can naturally generalize this integration to a $d$-dimensional rule, building the Cartesian product of the discretization points. Using $m+1$ points in each dimension, this requires $(m+1)^d$ points in total.

For the higher dimensional Simpson's rule (in $\mathbb{R}^d$) the error estimate is

$$
\left| \tilde{I}_{\text{Smps}}(f; a, b, (m+1)^d) - \int_{[a,b]^d} f(x) \, dx \right| \leq O(h^4)
$$

where $h = \frac{b - a}{m}$ using a total of $n = (m + 1)^d$ sample points. That is, in terms of the number of function evaluations we have

$$
\left| \tilde{I}_{\text{Smps}}(f; a, b, (m+1)^d) - \int_{[a,b]^d} f(x) \, dx \right| \leq O\left(n^{-\frac{4}{d}}\right)
$$


### **CLASSICAL INTEGRATION IN HIGHER DIMENSION**

Consider an integration rule $I$ that uses $m$ evaluation points $\Lambda_m := \{z_0, \dots, z_{m-1}\} \subset [0, 1]$ for an integrand $h$ to approximate the 1-dimensional integral

$$
\int_0^1 h(x) \, dx \approx I(h; m).
$$

The integration rule can be generalized to $d$ dimensions via iterated integrals. Let $g^{(d)} := f$ and define by backward induction for $k = d, d-1, d-2, \dots, 1$.

For a fixed $x^{(k-1)} \in [0,1]^{k-1}$ the function

$$
x_k \mapsto g^{(k)}(x^{(k-1)}, x_k)
$$

is a function on $[0,1]$. We then define

$$
\dot{g}^{(k-1)}(x^{(k-1)}) := I(g^{(k)}(x^{(k-1)}, \cdot); m) \approx \int_0^1 g^{(k)}(x^{(k-1)}, x_k) \, dx_k.
$$

> For a function f: $[0, 1]^d \to \mathbb{R}$ this requires to evaluate the function on the Cartesian product $\Lambda_{m}^d. That is, n = m^d$ evaluation points
