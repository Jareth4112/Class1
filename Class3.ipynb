## A solution for the problem we could use the **KAHAN SUMMATION ALGORITHM**
- Calculates the sum of an array of numbers such that the error is O(1)(not acumulative with n), and independent of the lenght of the array  

```python
def kahanSum(realizations):
    sum = 0.0
    error = 0.0
    for i in range(len(realizations)): 
        value = realizations[i] - error # 0.1 - 0.0 /// 0.1
        newsum = sum + value # 0.0 + 0.1 /// 0.1 + 0.1 = 0.2
        error = (newsum - sum) - value # (0.1 - 0.0) - 0.1 = 0.0 /// (0.2 - 0.1) - 0.1
        sum = newsum # 0.1 /// 0.2
    return sum/len(realizations), sum
```

```python
value = 0.1
numberOfValue = 10000000 # The error improves as the number of sizes is higher

#Create an array 
realization = np.full(numberOfValue, value)

a = kahanSum(realization)
print(a)
```

```python
value = 1
numberOfValue = 5 # The error improves as the number of sizes is higher

#Create an array 
realization = np.full(numberOfValue, value)

a, b = kahanSum(realization)
print(a, b)
```
