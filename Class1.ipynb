# PARTIAL DERIVATIVES


$
f'(x) = \frac{d}{dx} f(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}
$

```python
import sympy as sp
import numpy as np
```

```python
def derivativeanalytic(expr, var):
    """
    Calcula la derivada analítica de una expresión con respecto a una variable.
    
    :param expr: La función simbólica a derivar (por ejemplo, x**2 + 3*x).
    :param var: La variable con respecto a la cual se deriva (por ejemplo, x).
    :return: La derivada simbólica de la función.
    """
    return sp.diff(expr, var)

# Ejemplo de uso
x = sp.symbols('x')  # Definir la variable simbólica
f = sp.exp(x) # Definir la función simbólica

# Calcular la derivada
derivada = derivativeanalytic(f, x)
print("La derivada de f(x) es:", derivada)

# Evaluar la derivada en x = 0
valor_en_cero = derivada.subs(x, 0)
print("La derivada de f(x) evaluada en x = 0 es:", valor_en_cero)
```

```python
#Derivada aproximada
def derivativeapprox(func, x_val):
    """
    Calcula la derivada aproximada de una función en un punto usando diferencias finitas.
    
    :param func: La función a derivar (por ejemplo, lambda x: np.exp(x)).
    :param x_val: El punto donde se evalúa la derivada.
    :param shift: El pequeño desplazamiento para calcular la diferencia (por defecto 1e-5).
    :return: La derivada aproximada en el punto dado.
    """

    shift = 1e-5
    return ((func(x_val + shift) - func(x_val)) / shift)

f = lambda x: np.exp(x)

x_val = 0 #Evaluar la derivada en x=0
derivada_aprox = derivativeapprox(f, x_val)
print("La derivada aproximada de f(x) en x = 0 es:", derivada_aprox)
print("El error aproximado es:", abs(derivada_aprox - valor_en_cero))
```

**`El error *h* incrementa entre $10^-15$ y $10^-8$ `**
