# **FROM ACCEPTANCE-REJECTION SAMPLING TO WEIGHTED MONTE CARLO**
- The AR method is low when we throw away a lot of points, but if we consider the MC integration as an application, it's maybe not even necessary to throw away points. Every points may be used. 

Consider the approximation of $ E(h(X)) $ using Monte Carlo integration:

$$
\frac{1}{n} \sum_{i=1}^{n} h(x_i)
$$

where $ x_i $ is generated via acceptance-rejection sampling.

We can then replace $ h(x_i) $ with $ h(y_j) \cdot 1_{u_j \leq \frac{f(y_j)}{c g(y_j)}} $ and sum over all $ j $ until $ n $ points are accepted, that is:

$$
\sum_{j=1}^{m} 1_{u_j \leq \frac{f(y_j)}{c g(y_j)}} = n
$$

Therefore:

$$
\frac{1}{n} \sum_{i=1}^{n} h(x_i) = \frac{1}{m} \sum_{j=1}^{m} h(y_j) \cdot 1_{u_j \leq \frac{f(y_j)}{c g(y_j)}} = \frac{1}{m} \sum_{j=1}^{m} 1_{u_j \leq \frac{f(y_j)}{c g(y_j)}}
$$


$$
\begin{array}{|c|l|}
\hline
\textbf{Symbol} & \textbf{Meaning} \\
\hline
x & \text{Random variable of interest} \\
f(x) & \text{Target distribution (PDF of } x\text{)} \\
g(x) & \text{Proposal distribution (easy to sample from)} \\
c & \text{Scaling constant such that } f(x) \leq c \cdot g(x) \text{ for all } x \\
u & \text{Uniform random number in } [0,1] \\
y_j & \text{Sample from } g(x) \\
u_j & \text{Corresponding uniform random number} \\
x_i & \text{Accepted sample (i.e., } y_j \text{ such that } u_j \leq \frac{f(y_j)}{c g(y_j)} \text{)} \\
m & \text{Number of accepted samples (final sample size)} \\
n & \text{Number of total samples proposed (before rejection)} \\
h(x) & \text{Function of interest (e.g., payoff, } \log(x), x^2, \text{ etc.)} \\
E(h(X)) & \text{Expected value under target distribution } f(x) \\
\hline
\end{array}
$$

---
In the sum, we multiply by 0 or 1 depending on the sampled values $ y_j $ and $ u_j $.

Consider $ y $ being fixed: instead of setting the valuation to 0 for some $ u $'s, we could also **`consider accounting for their relative frequency by a weight.`**

Note that the sums are Monte Carlo approximations of a two-dimensional integral, namely:

$$
\frac{1}{m} \sum_{j=1}^{m} h(y_j) \cdot 1_{u_j \leq \frac{f(y_j)}{c g(y_j)}} \xrightarrow{m \to \infty} \int_{-\infty}^{\infty} \int_{0}^{1} h(y) \cdot 1_{u \leq \frac{f(y)}{c g(y)}} \cdot g(y) \, du \, dy
$$

$$
= \int_{-\infty}^{\infty} h(y) \left( \int_{0}^{1} 1_{u \leq \frac{f(y)}{c g(y)}} \, du \right) g(y) \, dy
$$

$$
= \frac{1}{m} \sum_{j=1}^{m} h(y_j) \cdot R(y_j)
$$

where the $\int_{0}^{1} du$ part can be integrated analytically. We have:

$$
R(y) = \int_{0}^{1} 1_{u \leq \frac{f(y)}{c g(y)}} \, du = \frac{1}{c} \cdot \frac{f(y)}{g(y)}
$$

>___

To compute $R(y)$, we evaluate the following integral:

$$
R(y) = \int_0^1 1_{u \leq \frac{f(y)}{c g(y)}} \, du
$$

The indicator function $1_{u \leq \frac{f(y)}{c g(y)}}$ is equal to 1 when $u \leq \frac{f(y)}{c g(y)}$, and 0 otherwise.

So we are integrating the value 1 over the interval from $u = 0$ up to $u = \frac{f(y)}{c g(y)}$, assuming this fraction lies in $[0,1]$.

Therefore, the result of the integral is just the **length** of the interval from $0$ to $\frac{f(y)}{c g(y)}$:

$$
R(y) = \frac{f(y)}{c g(y)}
$$

This is because:

$$
\int_0^1 1_{u \leq \alpha} \, du = \begin{cases}
\alpha, & \text{if } 0 \leq \alpha \leq 1 \\
1, & \text{if } \alpha > 1 \\
0, & \text{if } \alpha < 0
\end{cases}
$$

In our case, we typically ensure that $c$ is chosen such that:

$$
\frac{f(y)}{c g(y)} \leq 1
$$

for all $y$, so:

$$
R(y) = \frac{f(y)}{c g(y)}
$$

---
So instead of the Monte Carlo integral:

$$
\frac{1}{n} \sum_{i=1}^{n} h(x_i) = \frac{1}{m} \sum_{j=1}^{m} 1_{u_j \leq \frac{f(y_j)}{C g(y_j)}} \cdot h(y_j)
$$

we may consider (using all $ y_i $). Setting $ w(y) := C R(y) = \frac{f(y)}{g(y)} $, this is equivalent to:

**Equation (9):**

$$
\frac{1}{\frac{1}{m} \sum_{i=1}^{m} R(y_i)} \cdot \frac{1}{m} \sum_{i=1}^{m} h(y_i) R(y_i)
$$

**Equation (10):**

$$
\frac{1}{\frac{1}{m} \sum_{i=1}^{m} w(y_i)} \cdot \frac{1}{m} \sum_{i=1}^{m} h(y_i) w(y_i)
$$

The expressions (9) and (10) will also converge to $ E(h(X)) $ (proof follows).

Since

$$
\frac{1}{m} \sum_{i=1}^{m} w(y_i) \rightarrow \mathbb{E}(w(Y)) = \int_{\mathbb{R}} \frac{f(y)}{g(y)} g(y) \, dy = 1,
$$

the weighted average above is similar to approximating $\mathbb{E}(X)$ by

$$
\frac{1}{n} \sum_{i=1}^{n} h(y_i) w(y_i)
$$

Now,

$$
\frac{1}{n} \sum_{i=1}^{n} h(y_i) w(y_i)
$$

is the Monte Carlo approximation of

$$
\mathbb{E}(h(Y) w(Y)) = \int_{-\infty}^{\infty} h(y) \cdot w(y) \cdot g(y) \, dy. \quad \text{(11)}
$$

With $ w = \frac{f}{g} $, we see that this is a Monte Carlo approximation with weights $ w(y_i) $ approximating

$$
\int_{-\infty}^{\infty} h(y) \cdot w(y) \cdot g(y) \, dy = \int_{-\infty}^{\infty} h(y) \cdot f(y) \, dy = \int_{-\infty}^{\infty} h(x) \cdot f(x) \, dx = \mathbb{E}(h(X))
$$

- Remember: $ w(y) := C R(y) = \frac{f(y)}{g(y)} $

---
#### **Definition 58 (Weighted Monte-Carlo Approximation / Likelihood Ratio):**

Let $ w = \frac{f}{g} $, where $ f $ is the density of $ X $ and $ g $ is the density of $ Y $. Assume:

$$
g(y) = 0 \implies f(y) = 0
$$

Let $ y_i $ denote samples drawn from $ Y $. Then we define:

$$
\frac{1}{n} \sum_{i=1}^{n} h(y_i) w(y_i)
$$

as a **weighted Monte-Carlo approximation** of $ \mathbb{E}(h(X)) $. The weight $ w = \frac{f}{g} $ is called the **likelihood ratio**.

---

**Interpretation (of Monte Carlo integration with Acceptance-Rejection Sampling as Weighted Monte Carlo):**

A Monte Carlo integral with a sequence generated by acceptance-rejection can be interpreted in two or one dimension:

$$
\mathbb{E}(h(Y) \mid \{AR(U, Y) = 1\}) = \frac{\mathbb{E}(h(Y) \cdot AR(U, Y))}{\mathbb{E}(AR(U, Y))} = C \cdot \mathbb{E}(h(Y) \cdot AR(U, Y))
$$

$$
= \int_{[0,1]^2} h(G^{-1}(v)) \cdot C \cdot AR(u, G^{-1}(v)) \, d(u, v)
$$

$$
= \int_{0}^{1} h(G^{-1}(v)) \cdot C \cdot R(G^{-1}(v)) \, dv
$$

$$
= \int_{0}^{1} h(G^{-1}(v)) \cdot w(G^{-1}(v)) \, dv = \mathbb{E}(h(Y) \cdot w(Y))
$$
