# **GENERATING DRAWINGS OF OTHER DISTRIBUTIONS**
**Rembember the Monte Carlo steps**:
1. **Define the problem:** What do you want to analyze, simulate, estimate or solve ?
2. **Model the problem in an probabilistic way:** Model the problem in terms of random variables and their distributions.
3. **Generate random numbers`(SAMPLING)`:** Draw randon numbers from the distributions defined in the previos step. 
4. **Evaluate the object function:** For each random sample, evaluate the objective function.
5. **Average the results:** Combine the individual outcome to obtain the desired estimate value.
6. **Estimate the error:** Calculate a measure of the uncertainty in the estimate. (confindence intervals, standard deviation, variance, etc.)


##### **MOTIVATION:**
- The MC method approximates the expectation or a random variable by a sequence of i.i.d. randdom variables. 
- Let $X_i$ be a sequence of i.i.d. random variables with expectation $E[X]$. The MC method approximates $E[X]$ by the sample mean:
$$
\frac{1}{n} \sum_{i=1}^{n} X_i \approx E[X] \space \text{P-almost surely as} \space n \to \infty
$$

- The same hodls for $Zi:= f(Xi) \text{ and } Z:=f(X)$

$$
\frac{1}{n} \sum_{i=1}^{n} f(X_i) \approx E[f(X)] \space \text{P-almost surely as} \space n \to \infty
$$

> Until now, we have only used the uniform distribution to generate random numbers X to the integration of f on $[0, 1]^d$. 

**NOW IF WE ARE INTERESTED IN:**
$$
E[f(X)]
$$

- and X has a different distribution from the uniform, for example, a normal, log--normal, exponential, etc.
- In this case, we are interested in the following Monte Carlo:
$$
E[f(X)] \approx \frac{1}{n} \sum_{i=0}^{n-1} f(X_i) \approx \frac{1}{n} \sum_{i=0}^{n-1} f(X_i[\omega]) = \frac{1}{n} \sum_{i=0}^{n-1} f(x_i)
$$

- where $f(X_i[\omega])$ is the function evaluated at a specific realization of the random variable $X_i$

> So we are interested in generating sequence of random numbers $x_i = X_i[\omega]$, being drawings of X, where X has a given distribution (possibly different from the uniform distribution).

### **REALLY IMPORTANT EXAMPLE(UNIVERSAL PRICING THEOREM):**
- The value of a derivative is given as:

$$
V(t) = E^{Q^{N}}(V(T) \cdot \frac{N(t)}{N(T)}|\mathcal{F}_t)
$$

- $V(t)$ is the fair value of the derivative at time t
- $V(T)$ is the value of the derivative at maturity T, 
- $N(t)$ is the number of units of the underlying asset at time t, and 
- $N(T)$ is the number of units of the underlying asset at maturity T.
- $\mathcal{F}_t$ is the filtration at time t, which contains all the information available at time t.
- $Q^{N}$ is the risk-neutral measure associated with the numeraire $N$.

For example for a call option on the underlying $S(T)$, that is, the payoff:
$$
V(T) = \max(S(T) - K, 0)
$$
- $S(T)$ is the price of the underlying asset at maturity T, and
- $K$ is the strike price of the option.

valued under the terminal measure (i.e., $\frac{N(t)}{N(T)} = P(T;t)$) we have:
$$
V(t) = E^{Q^{N}}(\max(S(T) - K, 0) \cdot P(T;t)|\mathcal{F}_t) = E^{Q^{N}}(g(S(T)) \cdot P(T;t)|\mathcal{F}_t)
$$
Where:
- g(x) = $\max(x - K, 0)$ is the discounted payoff function. 

So the underlying asset $S(T)$ has to be modeled that is usually done by the Black-Scholes model, which has a log-normal distribution, that is:
$$
\log(S(T)) = \log(S(0)) + \bar{r}T - \frac{1}{2}\bar{\sigma^2} T + \bar{\sigma} {W^{\mathbb{Q}^{\mathbb{N}}}}(T)
$$
- Where $W^{\mathbb{Q}^{\mathbb{N}}}$ is a Brownian motion under the risk-neutral measure $\mathbb{Q}^{\mathbb{N}}$
    - Also is a random variable with mean 0 and standard deviation $\sqrt{T}$, that is:
    $$
    S(T) = S(0) \cdot \exp{(\bar{r} - \frac{1}{2}\bar{\sigma^2})T + \bar{\sigma}\sqrt{T}X} = h(X)
    $$
Where
- $X$ is a standard normal random variable, that is, $X \sim \mathcal{N}(0, 1)$
- S(0), $\bar{r} \space and \space \bar{\sigma}$ are model parameters.

**FINALLY**
- The value of the call option can be written as:
$$
V(t) = E^{Q^{N}}(g(h(X)) \cdot P(T;t)|\mathcal{F}_t) = E^{Q^{N}}(g(S(T)) \cdot P(T;t)|\mathcal{F}_t)
$$

Where
- g defines the financial product (the discounted payoff function)
- h defines the model for S 
- X is the stochastic driver (here a standard normal random variable), the driver that creates the uncertainty in the model *(risk in the stock and then in the payoff)*.

Assuming $\mathcal{F}_t = \{\emptyset, \Omega\}$ and setting `f(X):= g(h(x))` we have:
$$
V(t) = E^{Q^{N}}(f(X))
$$
- So the financial product valuation is an application of the Monte Carlo method, with X being a normal distributed random variable. 

## **INVERSION OF THE DISTRIBUTION FUNCTION**
- Robust and efficient method to generate random numbers from a distribution is the inversion of the distribution function. Meanwhile, it's necessary to have the distribution function $F_X$ of the random variable $X$.
- Given the distribution function $F_X$ we can create $F-distributed$ random numbers from uniform random numbers by "inverting" the function $F$. In general it's not necessary that $F$ is strictly *monotone*. 
- Remember $F(x) = P(X \leq x)$ is the distribution function of the random variable $X$.
- The probability density is the derivative of the $F$, $\phi(x) = \frac{\partial F}{\partial x}$ 
- *We therefore define a generalized inverse of $F$ (see Figure 22):*

**Definition 22 (Inverse of the Distribution Fuction):**

Let $F:\mathbb{R} \to [0, 1]$ denote a distribution function, i.e. $F(x):=P(X \leq x)$

So the inverse fo the distribution function, $F^{-1}$ is defined as:

$$
F^{-1}(y) := \inf\{x | F(x) \geq y\}, where \space y \in [0, 1]
$$

where:
- $\inf$ is the infimum, the greatest lower bound of the set.

> The set of all values $x$ where $F(x)$ is larger than or equal to $y$ is called the *level set* of $F$ at $y$.

**REMARK 49 (Distribution Function):**
- I need the following propertiy the distribution function is right-continuous, and hence i can change the *infimum* to a *minimum* in the definition of the inverse of the distribution function.
- We can prove that as follows:
$$
\Leftrightarrow \lim_{\epsilon \searrow 0} F(x + \epsilon) = F(x), \space \text{since for } \epsilon > 0
$$
$$
F(x+\epsilon) - F(x) = P(X \leq x + \epsilon) - P(X \leq x) = P(X \in (x, x + \epsilon]) \to P(\emptyset) = 0 \text{ for } \epsilon \searrow 0
$$

- The range of $P(X \leq x + \epsilon)$ is $(-\infty, x+\epsilon]$
- The range of $P(X \leq x)$ is $(-\infty, x]$ and hence the difference is the probability of the interval $(x, x + \epsilon]$.


#### **`Lemma 50 (Random Variable for a given distribution Fuction):`**
- Finally in inverse of the distributio function, we need this lemma.
- Let $U$ be a uniform distributed random variable on (0, 1), and $F$ a distribution function. Then
$$
X := F^{-1}(U)
$$
has distribution function $F$. If $F: \mathbb{R} \to (0, 1)$ is continuous then $F(X)$ is uniform (0, 1).  

> **Applying the inverse of the distribution function to a uniform distributed random variable generates an $F$ distributed random variable**

> **Given now a sequence of uniform drawings I just apply the inverse and I have a sequence of $F$ distributed drawings of drawings of this x**

> **Given a sequence of uniform samples, I can apply the inverse function to obtain a sequence of random variables distributed according to $F$.**

- **Proof:**

$$
X \text{ has distribution function } F \Leftrightarrow P(X \leq x) = F(x).
$$

$$
\text{We first show that} \\
** F^{-1}(y) \leq x ** \Leftrightarrow F(x) \geq y. \tag{7}
$$

$$
\text{To show (7): From the right-continuity of } F \text{ the } \inf \text{ in Definition 48 is attained. Then}
$$

$$
F^{-1}(y) \leq x \overset{\text{Def. } F^{-1}}{\Leftrightarrow} ** \min \{ \xi \mid F(\xi) \geq y \} \leq x **
$$

$$
\text{Since } \xi \leq x \Rightarrow F(\xi) \leq F(x) \text{ we have}
$$

$$
\min \{ \xi \mid F(\xi) \geq y \} \leq x \Rightarrow x \in \{ \xi \mid F(\xi) \geq y \} \Rightarrow F(x) \geq y,
$$

$$
\text{and}
$$

$$
F(x) \geq y \Rightarrow x \in \{ \xi \mid F(\xi) \geq y \} \Rightarrow \min \{ \xi \mid F(\xi) \geq y \} \leq x,
$$

$$
\text{Together this is (7).}
$$

#### **`Showing Lemma 50`**
- The uniform distributed random variable $U$ has distribution function:
$$
G(x) =
\begin{cases}
0 & \text{for } x < 0, \\
x & \text{for } 0 \leq x \leq 1, \\
1 & \text{for } x > 1.
\end{cases}
$$

- For $u \in [0, 1]$ we have $P(U \leq u) = G(u) = u$
- Let $x \in \mathbb{R}$. Then we have
$$
P(X \leq x)
\overset{\text{Def. } X}{=}
P(F^{-1}(U) \leq x)
\overset{(7)}{=}
P(U \leq F(x))
\overset{U \text{ equi-dist.}}{=}
F(x).
$$

$
\text{Thus, } X \text{ has distribution function } F.
$


#### **Remark 51 (Inversion of the ICDF for a Quasi Monte Carlo Approximation):**
- The method also applies to the Quasi Monte Carlo method, where the uniform random variable is replaced by a low-discrepancy sequence.
- For a given function f: $\mathbb{R} \to \mathbb{R}$ we find from ** $x = F^-1(u)$ ** that ** $du = \phi(x) dx$ **, where $\phi(x) = F'(x)$ is the density of X, ad thus:

$$
\mathbb{E}(f(X)) = \int_{\mathbb{R}} f(x) \phi(x) \, dx = \int_{[0,1]} f(F^{-1}(u)) \, du
$$

- Where:
    - $\phi(x)$ is the probability density function of the random variable $X$,
    - $f(x)$ is the funcyion of interest that we want to evaluate the expectation of,
    - $F^{-1}(u)$ is the inverse of the distribution function of $X$,
    - $u$ is a uniform random variable on the interval $[0, 1]$.
---

$
\text{So in this setup the Koksma–Hlawka inequality reads}
$

$$
\left| \frac{1}{n} \sum_{i=1}^{n} f(x_i) - \mathbb{E}(f(X)) \right| \leq V(f \circ F^{-1}) D^*(u_1, \ldots, u_n),
$$

$
\text{where } V(h) \text{ is the variation of the function } h, \text{ and } \{u_1, \ldots, u_n\} \subset [0, 1] \text{ is a given sequence,}
$

$$
\text{and } x_i = F^{-1}(u_i) \text{ is the transformed sequence.}
$$

where: 
- $V(f \circ F^{-1})$ is the variation of the function $f$ composed with the inverse of the distribution function $F^{-1}$.
- $D^*(u_1, \ldots, u_n)$ is the star discrepancy of the sequence $\{u_1, \ldots, u_n\}$.

### **APPLICATION CONSTRAINS:**
- We have an inverse of this distribution function *analytically*, so we can use it in the computer. Often the close formula for the distribution function and the inverse of the distribution function (ICDF) does not exist. 
- Meanwhile, there are very fast approximations for the inverse of the distribution functions (ICDF) 
- If the approximation error is larger than the machine precision, it may be tempting to improve it using a numerical method (e.g., Root finder for the inversion of the df, but it may lead to undesired effects and is rarely needed)

- In many applications, an approximation error in the ICDF is usually easy to control. For example, in a derivative valuation model, it can be checked as a calibration error of the model. In some situations, the error can be controlled (reduced) as a part of calibration of the model parameters.

- Improving the accuracy of the ICDF by a numerical root finder can result in a more accurate model, *but discontinuos function*, which can produce unstable results when considering finite differences or results generated from such a model.

### **WARNING WITH THE DOMAIN OF THE INVERSION:**
- It's important to check the domain of the inverse of the ICDF and the behavior of the transformation and its specific implementation in the context of the random number generators.
- For example, the inverse of the inverse distribution function of the normal distribution is defined on the open set (0, 1). The values 0 and 1 map to $-\infty$ and $+\infty$, respectively, which can lead to numerical issues if not handled properly.

- Random number generators usually generate random sequence on the set [0, 1) (0 included, 1 exluded) (note: some will never generate a double 0.0). The first element to the Sobol sequence is 0.  
>---

> You should avoid using 0 and 1 as inputs to an ICDF (such as for the normal distribution), because they can return ±∞ and cause numerical errors—especially if your random number generator (like Sobol) can produce 0.
