# **Generating Drawings of Other Distributions**

- First, we gonna start with the **normal distribution**.
- For a standard normal distributed random variable X, the density is known analitcally as:
$$
\phi_{std. norm.}(x) = \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}}
$$

but for the distribution function
$$
F_{std. norm.}(x) = \int_{-\infty}^{x} \phi_{std. norm.}(\xi) d\xi
$$

an analytic formula for the inverse $F^{-1}$ is not known.

However, there exists a piecewise rational approximation for which the approximation-error is smaller than the *double floating point* machine precision (less than the rounding error of the machine)- that is - the implementation of the inverse function is exact with respect to IEEE 754 rounding.

## **RECALL: LEMMA 36 - FLOATING POINT NUMBERS**
Let $0 \leq x < m$ be an integer. Let $p = q + 1$ denote the precision as in Definition 1.  
If $e_{\text{min}} \leq -(q + 1)$ and $m = 2^{q+1}$ then $\frac{x}{m}$ is a floating point number, i.e., no rounding is performed.

**Proof**:  
- For $x = 0$, $z = \frac{x}{m} = 0$ is a denormalized floating point number.  
- For $x > 0$ we show that $x$ is a normalized floating point number.  

- Let $k \geq 0$ be such that  
$$
2^k \leq x < 2^{k+1}
$$

- Then  
$$
0 \leq x - 2^k < 2^k
$$

- We have:
$$
\frac{x}{m} = \frac{2^k + (x - 2^k)}{m}
= \left(1 + \frac{x - 2^k}{2^k}\right) \frac{2^k}{m}
= \left(1 + \frac{(x - 2^k) 2^{q - k}}{2^q}\right) 2^{k - (q + 1)}
= \left(1 + \frac{c}{2^q}\right) 2^e
$$

- with  
$$
c = (x - 2^k) 2^{q - k} \quad \text{and} \quad e = k - (q + 1)
$$

- Since  
$$
0 \leq x - 2^k < 2^k \Rightarrow 0 \leq c < 2^q, \quad \text{and} \quad k \geq 0 \Rightarrow e \geq e_{\text{min}},
$$

- we find that $x$ is a normalized floating point number.
