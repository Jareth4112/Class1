# **TIME DISCRETIZATION OF STOCHASTIC PROCESSES**
Random variables parameterized by time
$$
X(t) = \mu(X(t), t) dt + G(t, X(t))dW
$$
- Drift part(`determinist tendency, without uncertainty`): $\mu(X(t), t) dt$
- Diffusion part(`randomness of the process, stochastic part`): $G(t, X(t))dW$

So we move from the infinitesimal changes $X(t)$ to the finite changes $X(t + \Delta t) - X(t)$

$$
\Delta \tilde{X}(t_i) = \mu(\tilde{X}(t_i), t_i) \Delta t_i + G(t, \tilde{X}(t_i))\Delta W(t_i)
$$

- - $\tilde{X} = \text{ An approximation of X}$
- $\Delta \tilde{X}(t_i) = \tilde{X} (t_{(i+1)}) - \tilde{X} (t_i)$
- $\Delta t_i = t_{i+1} - t_i$
- $\Delta W(t_i) = W(t_{(i+1)}) - W(t_i)$

### `Itô's Lemma:`

If $X(t)$ follows:
$$
dX(t) = a(X,t) \, dt + b(X,t) \, dW(t),
$$
and $f$ is $C^2$, then:
$$
df = \frac{\partial f}{\partial t} \, dt
+ \frac{\partial f}{\partial x} \, a(X,t) \, dt
+ \frac{\partial f}{\partial x} \, b(X,t) \, dW(t)
+ \frac{1}{2} \frac{\partial^2 f}{\partial x^2} \, b(X,t)^2 \, dt.
$$


Application: $f(S) = \log S$, Black–Scholes model

$$
dS = \mu S\, dt + \sigma S\, dW
$$
- $f'(S) = \frac{1}{S}$
- $f''(S) = -\frac{1}{S^2}$
- $a(S) = \mu S$
- $b(S) = \sigma S$

Substituting into Itô's Lemma:
$$
d\log S = \mu\, dt + \sigma\, dW - \frac{1}{2} \sigma^2\, dt
$$
$$
\Rightarrow d\log S = \left( \mu - \frac{1}{2} \sigma^2 \right) dt + \sigma\, dW
$$

**`Key:`** The drift is reduced by $\frac{1}{2}\sigma^2$ due to the variance of the Brownian motion.

## **TIME DISCRETIZATION OF TIME DISCRETE ITÔ PROCESSES**
We discuss the Monte-Carlo simulation of time-discrete Itô-processes for the example of the valuation of an Asian option (the product) under the Black-Scholes model (the model).

In more general applications, $V(T)$ may have a more complex dependence on $\mathcal{F}_T$-measurable events.

For example, the time $T$ value of an **Asian option** on the stock $S$ is given by:

$$
V(T) = \max \left( \frac{1}{m} \sum_{i=1}^{m} S(T_i) - K,\ 0 \right), \quad T_1 < T_2 < \ldots < T_m =: T.
$$

In this context, we have to generate Monte Carlo samples (drawings) of the $m$-dimensional random vector:

$$
X = \left( S(T_1),\ S(T_2),\ \ldots,\ S(T_m) \right).
$$

<img src="image-31.png" alt="Asian Option" width="600"/>
</p>


> The payoff depends not an single RV, but a vector of RVs X, that represent the time evolution of the stock price. So, this is a problem of high dimension with six RV. 

In other words, we sample a (time-discrete) path $\omega$ of the stochastic process $S$:

$$
i \mapsto S(T_i, \omega), \quad i = 0, 1, 2, \ldots, m.
$$

In this context, we call the sampling of a path of a stochastic process **Monte-Carlo simulation** (synonym to Monte-Carlo sampling).

Assuming (for simplicity) that the numéraire ratio $N(t)/N(T)$ is constant, we see that the valuation of the Asian option corresponds to an expectation:

$$
V(t) = \mathbb{E}^{\mathbb{Q}^N} \left[ f(X) \mid \mathcal{F}_t \right] = \mathbb{E}^{\mathbb{Q}^N} \left[ f(S(T_1), S(T_2), \ldots, S(T_m)) \mid \mathcal{F}_t \right].
$$

The Monte-Carlo approximation corresponds to an $m$-dimensional Monte-Carlo integral.

### **MODEL**
The random variables $S(t)$ may be defined through a model, which gives each $S(T_j)$ as a function of a high-dimensional random vector (and depending on the model, even higher dimensional).

Consider the Black-Scholes model:

$$
dS(t) = \mu S(t)\, dt + \sigma S(t)\, dW(t).
$$

`From Itô's Lemma with` $Y := \log(S)$ we find:

$$
d\log(S(t)) = \left( \mu - \frac{1}{2} \sigma^2 \right)\, dt + \sigma\, dW(t).
$$

Let $0 = t_0 < t_1 < \cdots < t_n = T$ denote a given time discretization.  
Integrating over $\int_{t_i}^{t_{i+1}}$ we get:

$$
\log(S(t_{i+1})) = \log(S(t_i)) + \int_{t_i}^{t_{i+1}} d \log{(S(t))}= \left( \mu - \frac{1}{2} \sigma^2 \right)(t_{i+1} - t_i) + \sigma (W(t_{i+1}) - W(t_i)).
$$

- $(W(t_{i+1}) - W(t_i)) \sim \mathcal{N}(0,(\Delta t_i))$

Taking the exponential:

$$
S(t_{i+1}) = S(t_i) \cdot \exp \left[ \left( \mu - \frac{1}{2} \sigma^2 \right)(t_{i+1} - t_i) + \sigma (W(t_{i+1}) - W(t_i)) \right].
$$

---
#### Black-Scholes Model

For Brownian increments $\Delta W(t_i) := W(t_{i+1}) - W(t_i)$, we have (more or less by definition) that they are independent and normally distributed with mean 0 and standard deviation $\sqrt{t_{i+1} - t_i}$, that is:

$$
W(t_{i+1}) - W(t_i) = \Delta W(t_i) = \sqrt{t_{i+1} - t_i} \cdot Z_i
$$

where $Z_i \sim \mathcal{N}(0, 1)$ are independent and identically distributed (i.i.d.) random variables.

In summary, we have: In the Black-Scholes model for an asset $S$, the time $t_{i+1}$ asset value $S(t_i+1)$ satisfies:

$$
S(t_{i+1}) = S(t_i) \cdot \exp{((\mu - \frac{1}{2} \sigma^2)(t_{i+1} - t_i) + \sigma \sqrt{t_{i+1} - t_i} Z_i)}
$$
$$
s(t_0) = S_0 \in (0, \infty)
$$

where $Z_i \sim \mathcal{N}(0, 1)$ are i.i.d.

##### **Summary: Simulation and Expectation in Asian Options**

- The vector of asset prices $(S(t_1), \ldots, S(t_n))$ is a function of the i.i.d. random variables $(Z_0, \ldots, Z_{n-1})$.
- The vector $(Z_0, \ldots, Z_{n-1})$ can be defined using the ICDF method from i.i.d. uniform random variables $(U_0, \ldots, U_{n-1})$ on $[0,1]^n$.

For the example of the Asian option above, it is sufficient to simulate only the product dates $T_i$, i.e., choose $n = m$ and $t_i = T_i$. We observe that:

- The expectation of the Asian option value $V(T)$ is (at least) an $m$-dimensional integral over $[0,1]^m$.

For more complex models, the number of time steps $n$ may need to be large — for example, to reduce the time-discretization error of numerical schemes.

### **Remark 61 — Monte Carlo Simulation of Stochastic Processes is Inherently High-Dimensional**

If we need to simulate the random variables $S(t_1), \ldots, S(t_n)$ and calculate an expectation of a function depending on those, then we are dealing with an $n$-dimensional integration of a function of the $n$-dimensional random vector:

$$
(S(t_1), \ldots, S(t_n)).
$$

The number $n$ is simply the number of time steps in the time-discretization.  
So, for example, using 100 time-discretization steps results in a 100-dimensional integration.

However, even if we only need to evaluate the expectation of a function on $S(T_m)$, this can still be a high-dimensional integral if $S(T_m)$ is constructed from smaller time steps.
