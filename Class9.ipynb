# **MONTE CARLO (INTERPRETATION AND INTUITION) CONVERGENCE IN HIGH DIMENSION**
- Recall that in the Monte Carlo integration formula, the error was given by:
$$
|\frac{1}{n} \sum^{n}_{i=1} f(X_i) - \int_{[0, 1]} f(X) dx|
$$
- that error being below $\sigma/\delta^{1/2} \cdot n^{1/2}$ is larger than $1-\delta$, where $\sigma^{2} = Var(f(X))$.
- Also, recall the convergence rate wich is:
$$
1/\sqrt{n}
$$

- And all of these results were based only in probability.
- So, at the first, the result appears to be unsatisfactory in the following sense:
    - The error estimate holds only in probabilistic sense. If we use one Monte-Carlo drawing $w$, then we have a non zero probaility that the error estimate doest not hold for this drawing.

- We can not fix this by using more points because either using more points improves the error stimate or improves the probability.
    - Improving the probability: $1-\delta \to 1-\delta/2$ 
    - Improving the accuracy by a factor of $1/\sqrt{2}$ by increasing n 




## Recall the curse of the dimensionality in the Monte Carlo integration

### Remark 28 (Curse of Dimensionality)

Compare the error estimate of the **Simpson’s rule**  
to that of a **Monte-Carlo integral**, where we have (note: in probability)

$$
\left| \tilde{I}_{\text{Smps}}(f; a, b, n) - \int_{[a,b]^d} f(x)\, dx \right| \leq C_{\text{Smps}}(f, a, b) \cdot n^{-\frac{4}{d}}
$$

$$
\left| \tilde{I}_{\text{MC}}(f; a, b, n) - \int_{[a,b]^d} f(x)\, dx \right| \leq C_{\text{MC}}(f, a, b) \cdot n^{-\frac{1}{2}}
$$

For large dimensions (e.g. $d > 8$), Monte-Carlo is much more efficient.  
Here $n$ denotes the number of function evaluations.

---

For both methods, the constants relate to the variability of the function. It is

$$
C_{\text{Smps}}(f, a, b) = C (b - a)^d \max_{\xi \in [a,b]^d} \sum_{i=1}^d \left| \frac{\partial^4 f}{\partial x_i^4}(\xi) \right|
$$

$$
C_{\text{MC}}(f, a, b) = \sqrt{ \frac{\text{Var}(f(X))}{\delta} }
$$


## **ILLUSTRATION HOW MONTE CARLO ACHIEVES IT CONVERGENCE RATE IN HIGHER DIMENSION**
- Suppose we want to calculate the integral of a function of two variables:
$$
f: [0,1]^2 \to \mathbb{R}
$$

- **`A classical rule is to interpret the integral as the iteration of one-dimensional integrals:`** for fixed $x_2$ integrate over $x_1$, then integrate over $x_2$:.

$$
\int_{[0, 1]^2} f(x_1, x_2) dx_1 dx_2 = \int_{0}^{1} (\int_{0}^{1} f(x_1, x_2) dx_1) dx_2
$$

$$
g(x_2)  = \int_{0}^{1} f(x_1, x_2) dx_1 \space \space \to \space \space \int_{[0, 1]^2} f(x) dx = g(x_2) dx_2
$$

$\int_{[0,1]^2} = \int_{0}^{1} \cdot \int_{0}^{1}$

- **`Applying a classical one dimensional integration rule:`** if the inner integration rule is always the same (i.e. while the integrand depends only on $x_2$, the rule is independent of $x_2$, then this corresponds to choosing the integration points $x_i = (x_{1, i},x_{2, i})$ as a cartesian product, that is $x_{1, i} = x_{1, i+m}$ (if we consider m integration points for the inner integral))

> Applying Simpsom rule or Trapezoid rule If we choose m points in $x_1$ and $x_2$ the neccessaty number of points is $m^2$ (cartesian product in each axis). So, if we had **d** dimensions, we would need $m^d$ points.

### **For Examples: The classical Riemman Sum**
- It chooses as integration points the **"center-points"** of intervals, i.e:
$$
\int_0^1 h(\xi) d\xi = \frac{1}{m} \sum_{i=0}^{m-1} h(\xi_i), \quad \xi_i = \frac{i + \frac{1}{2}}{m}, \quad i = 0,1,\dots,m-1
$$

- then the two dimensional integral is just

$$
\int_{[0,1]^2} f(x) dx = \frac{1}{n} \sum_{i=0}^{n-1} f(x_i), \quad x_i = (x_{1,i}, x_{2,i}), \quad i = 0,1,\dots,n-1
$$

- where $x_i$ runs through the *Cartesian product*

$$
\left\{ \frac{i + \frac{1}{2}}{m} \ \middle| \ i = 0, 1, \dots, m-1 \right\}
\times
\left\{ \frac{i + \frac{1}{2}}{m} \ \middle| \ i = 0, 1, \dots, m-1 \right\}
=
\left\{ x_i \ \middle| \ i = 0, 1, \dots, n-1 \right\}
$$

with \( n = m^2 \).

> You can see that the points are regular in the unit cube, but in the MC there are empty spaces by the randomness, and is not a bug, it's a feature 

- Consider the special case of a function f which doesn't depend on $x_2$. So we have $f(x_1, x_2) = h(x_1)$ (it only dependes on $x_1$). In this case the inner integration is always the same and it is approximated using m points. 

$$
\int_{0}^{1} (\int_{0}^{1} h(x_1) dx_1) dx_2
$$

- So we can move $x_2$ out to the integral 
$$
\int_{0}^{1} h(x_1) (\int_{0}^{1} dx_2)dx_1 = \int_{0}^{1} \cdot 1 dx_1 = \int_{0}^{1} h(x_1) dx_1
$$

- So you only need do the integrate once on $x_2$ because it doesn't change with $x_2$. So, the process become to an 1D integration with m points(for $x_1$). 

> This case does not break the curse of dimensionality - it simply avoids it because the effective dimensionality of the problem is reduced. But as soon as the function actually depends on all variables (which is common in many real problems), the curse is still as present as ever.

- `In contrast, Monte Carlo Integration chooses the evaluation points *randomly in every component* and the probability that a coordinate occurs twice is zero. `

- The price that we pay for the randomness are the gaps and clusters.

> So the advantages of using Monte Carlo are the following:
> - Simplicity of implementation:
>    - Very easy to apply, even in high dimensions.
>    - Just generate random numbers, evaluate the function, and sum the results.
> - Progressive accuracy improvement:
>    - You can increase accuracy simply by adding more function evaluations, without discarding previous ones.
>    - This is unlike classical numerical integration methods (like Simpson’s rule), where improving accuracy is more complex.
> - Independence from dimensionality:
>    - The complexity of the method and its convergence rate do not depend on the dimension of the problem.
>    - This makes Monte Carlo methods especially powerful in high-dimensional spaces.


>  **Monte Carlo integration uses random samples, which introduces some probabilistic error. However, you can remove the randomness by using deterministic sequences that are evenly spread over the domain — these are called low-discrepancy sequences (e.g., Sobol, Halton).
