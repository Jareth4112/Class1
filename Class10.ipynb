### **LINEAR CONGRUENTIAL GENERATORS**
- Classic random number generator, but are of poor quality. 
- A linear congruential generator generates the sequence:

$$
x_i := (a \cdot x_{i-1} + c) \mod m, \quad i = 1, 2, 3, \dots
$$

$$
\text{where } m \in \mathbb{N}, \quad a, c, x_0 \in \mathbb{N} \cup \{0\}, \quad 0 \leq a, c, x_0 < m^5
$$

- The initial value $x_0$ needs to be specified.
- The *mod* operator gives the residue of the division the first term by $m$.
- Generates a number between $0$ and $m-1$.

> **Note:** The quality of the random number generator depends on the choice of $a$, $c$, and $m$.
> Many build-in random number generators are linear congruential generators, that means they are produced by a linear function with a modulus(congruential) operation.


- **CASE MONTE CARLO**
- You need to be careful when using linear congruential generators in Monte Carlo simulations because they have undesired properties. 
- For example, $x_i$ be as LCG:
$$
y_i := (y_{i,1}, ..., y_{i, d})^T \in \mathbb{R}^d \space\space y_{i, j} = x_{i\cdot d+j} \space\space j = 1, \dots, d; i = 1, \dots, n

p = period 
$$
- Then $y_i$ lie on at most $m^{1/d}$ different hyperplanes

#### **LCG's PROPERTIES**
##### **1. PERIOD LENGHT**
- When the sequence repeats
- **Definition 31 (Period Length of a Random Number Generator)**

- Let $\{ x_i \mid i = 0,1,2,\dots \}$ be a random number sequence. Then

$$
\min\{ p \mid p > 0 \text{ and } \forall i : x_{i+p} = x_i \}
$$

- is called **period length** of the random number sequence.

- The smallest period possible under a random number generator is called *period length of the random number generator*.

---

- **Remark 32**

The period length of a linear congruential generator is at most $m$, i.e.,  
$x_{i+k} = x_i$ for some $k \leq m$.

### **MERSENNE TWISTER**
- Popular random number generator *for uniform distributed randon number* has a period length of $2^{19937} - 1$(for the first step it repeats after $2^{19937}$). Also, the random numbers are equidistruted in high dimensions (up to 623 dimensions, bad in the discretization).
- Is not an LCG, but shares some properties.
- While some classic LCG work on 32 bit word, an integer, or 48 a bit word; the Mersenne Twister operates on a 19937 bit word - this is 624 integers of 32 bits each.

### **RANDOM NUMBER GENERATOR: Uniform on [0, 1)**
- The LCG as defined above (specially in the part of the Mersenne Twister) generate integer numbers (more precise: bit sequence).
- If we want to generate a random number between [0, 1) and the generator generates uniforms integers x with 0 <= x < m, we just set z := x/m. 

> **Note:** So, if the integer is uniform on [0, 1), then z:= x/m is uniform on [0, 1>. However, there could be a rounding to a floating point number(**careful with these**)

- **LEMMA(Floating Point Numbers):**
- Let $0 \leq x \leq m$ be an integer. Let p = q + 1 denote the precision. If $e_min \leq -(q+1)$ and $m = 2^(q+1)$ then $x/m$ is a floating point number, i.e. no rounding is performed. 

- **PROOF:**
**Proof:**  
For $x = 0$, $x$ is a denormalized floating point number.  
For $x > 0$ we show that $x$ is a normalized floating point number.  
Let $k \geq 0$ be such that $2^k \leq x < 2^{k+1}$.  
Then $0 \leq x - 2^k < 2^k$.

We have:

$$
\frac{x}{m} = \frac{2^k + (x - 2^k)}{m} = \left(1 + \frac{x - 2^k}{2^k} \right) \frac{2^k}{m}
$$

$$
= \left(1 + \frac{(x - 2^k) 2^{q - k}}{2^q} \right) 2^{k - (q + 1)} = \left(1 + \frac{c}{2^q} \right) 2^e
$$

with

$$
c = (x - 2^k) 2^{q - k}, \quad \text{and} \quad e = k - (q + 1).
$$

Since $0 \leq x - 2^k < 2^k \Rightarrow 0 \leq c < 2^q$, and $k \geq 0 \Rightarrow e > e_{\min}$,  
we find that $x$ is a normalized floating point number. $\square$

> **Note:** So, $x/m$ is a floating point number that doesn't need to be rounded, only if m < 2^24 (In floating point numbers q = 23 and q + 1 = 24, In floating point doubles :q is 52, q+1 = 53 (2^26 + 2^27))
