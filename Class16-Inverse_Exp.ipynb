# **Generating Drawings of the Exponential Distribution**
#### **Definition 53 (Exponential Distributed Random Variable):**

Let $\lambda > 0$. The cumulative distribution function (CDF) of an exponentially distributed random variable is given by:

$$
F : \mathbb{R} \to [0, 1]
$$

$$
F(x) = 
\begin{cases}
1 - \exp(-\lambda x) & \text{if } x \geq 0 \\
0 & \text{if } x < 0
\end{cases}
$$

Where:

$$
P(X \leq x) = F(x) \mapsto \text{CDF}
$$

The probability density function (PDF) is obtained by differentiating $F(x)$:

$$
f'(x) = 
\begin{cases}
\frac{dF}{dx}(x) = \lambda \cdot \exp(-\lambda x), \quad & \text{if } x > 0 \\
\frac{dF}{dx}(0) = \lambda & \text{if } x = 0
\end{cases}
$$

#### **Inversion Method for (Homogenuos) Exponential Distribution:**
We may use the inversion method to generate an exponential distributed random variable from a uniform distributed random variable: If U is a uniform random variable then:

- Proof:
$$
F(x) = U \to 1 - \exp{(-\lambda x)} = U  \to \exp{(-\lambda x)} = 1 - U \mapsto x = - \frac{1}{\lambda} \log{(1-U)}
$$

> A significant interpretation, exponential distribution is used in the modelling of `default times` ($t: \mathbb{R} \to [0, \infty)$) until the occurrence of an event at a `stochastic time` $\tau$ where ($\tau: \mathbb{\Omega} \to [0, \infty)$) with $P(\tau \leq t) = 1 - \exp{(- \lambda t)}$: e.g. the credit default from a client, and the probability that the default has not  (`survival probability`) occurred yet is $\exp{(- \lambda t)}$. 


The distribution follows from the following model assumption that the *trigger* which leads to the default event is memoryless. That is, for $T_2 > T_1 \geq 0$ we have 
$$
P(\tau > T_2| \tau > T_1) = P(\tau > T_2 - T_1) \quad \text{(8)}
$$

- The probability that the event has **not occurred** by time $T_2$ given survival up to $T_1$ is equal to the probability that it **does not occur** in the interval of length $T_2 - T_1$.  
- This reflects the **memoryless property**: the probability depends only on the length of the interval, not on when it starts.

$$
P(\tau > T_2 \mid \tau > T_1) \overset{\text{def}}{=} \frac{P\left( \{ \tau > T_2 \} \cap \{ \tau > T_1 \} \right)}{P(\tau > T_1)}
$$

This is equivalent to
$$
P(\{\tau > T_2\}) = P(\tau > T_2 - T_1) \cdot P(\tau > T_1)
$$

Or 
$$
P(\tau > T_2 \mid \tau > T_1) = \frac{P(\tau > T_2)}{P(\tau > T_1)}
$$

> We need to check that the exponential distribution satisifies (8)

> Remember `the survival prob`: $\exp{(- \lambda t)}$.
> Using this $\exp{(- \lambda T_2)} \to P(\tau > T_2 \mid \tau > T_1) = \frac{\exp{(- \lambda T_2)}}{\exp{(- \lambda T_1)}} = \exp{(-\lambda(T_2 - T_1))}$
> 

##### **Conversely:**
If the trigger is memoryless, then this implies (considering times $T_i = \frac{i}{n} T$)
$$
P(\tau > T) = \prod{n}_{i=1} P(\tau > T/n) = P(\tau \geq T/n)^n
$$

<img src='image-19.png' alt='alt-text' widht=200 height=100>
</p>

From this we get
$$
P(\tau > T) = P(\tau > T/n)^n = \exp(\log(P(\tau > T/n)\cdot n)) = \exp{(\frac{\log{(P(\tau>T/n))}}{T/n}\cdot T)}
$$

and the limit 
$$
\lim_{n\to\infty}{(\frac{\log{(P(\tau>T/n))}}{T/n}\cdot T)} = -\lambda
$$
$$
P(\tau > T) = \exp(-\lambda T) \quad \text{Exponentially random variable}
$$

with
$$
\lambda := - \lim_{n \to \infty} \frac{P(\log{(\tau > T/n)})}{T/n} = - \lim_{n \searrow \infty} \frac{P(\log{(\tau> t)})}{t} = - \frac{\partial}{\partial t} P(\tau > t)\bigg|_{t = 0_+}
$$

$$
\frac{\partial}{\partial t} P(\log{(\tau> t)}) = \frac{[\frac{\partial P{(\tau>t)}}{\partial t}]}{P(\tau>t)} \quad \text{and for t = 0} \to P{(P(\tau > t))} = 1 \quad \text{so, } \frac{\partial P{(\tau>t)}}{\partial t}
$$

where:
- $P(\log{(\tau > T/n)})$ = Prob to survive small interval T/n

If $\tau$ models a *time* we see from

$$
P(\tau \leq t) = 1 - \exp(-\lambda t)
$$

or from the inversion of the distribution function

$$
\tau = -\frac{1}{\lambda} \log(1 - u)
$$

that the parameter $\lambda$ has the unit

$$
\frac{1}{\text{time}},
$$

i.e., it is a **frequency $(\lambda$ `how often do i get this event`), and $\frac{1}{\lambda}$ has the unit time **(`is the expected time of default`)**.

---

**Proof:**
, from the density of $\tau$ being

$$
\phi(t) = F'(t) = \lambda \exp(-\lambda t)
$$

we have

$$
E(\tau) = \int_0^{\infty} t \, \phi(t) \, dt = \frac{1}{\lambda},
$$

i.e., $\frac{1}{\lambda}$ is the expected time of the event and $\lambda$ then is the frequency of the events, given that we restart an (independent) experiment after the event has occurred.

### **Inhomogeneous Exponential Distributed Random Variable:**
#### Defintion 55: (Inhomogeneous Exponential Distributed Random Variable)
- Lambda is no longer assumed to be constant.

Let $\lambda : [0, \infty) \rightarrow [0, \infty)$ such that $\int_0^t \lambda(s) \, ds \rightarrow \infty \quad \text{as } t \rightarrow \infty.$

The distribution function of an *inhomogeneous exponential distributed random variable* is given by:

$$
F : \mathbb{R} \rightarrow [0, 1], \quad
F(t) =
\begin{cases}
1 - \exp\left(-\int_0^t \lambda(s) \, ds \right) & t \geq 0 \\
0 & t < 0
\end{cases}
$$

#### Inversion Inhomogeneous Exponential Distributed Random Variable:

Let  $\Lambda(t) := \int_0^t \lambda(s) \, ds.$

If the inverse $(\Lambda^{-1})$ is known, we know the inverse distribution function.

From  
$$
F(t) = 1 - \exp(-\Lambda(t))
$$ 
we get:

$$
\tau = F^{-1}(u) = \Lambda^{-1}(-\log(1 - u)).
$$

---

The inversion method can also be used to approximately sample an inhomogeneous exponential distribution. In practical applications, this is often sufficient or can be considered as an additional time-discretization step, since we don’t always have all the data needed to calibrate the model.


For this, let  
$$
0 = t_0 < t_1 < \cdots < t_n
$$  
be a time discretization, and define:

$$
\lambda_i := \frac{1}{t_{i+1} - t_i} \int_{t_i}^{t_{i+1}} \lambda(s) \, ds.
$$

Either we could consider only those \( \lambda \)'s that are constant on \( (t_i, t_{i+1}) \) *(a model assumption)*,  
or we could view the \( \lambda_i \)'s as the **piecewise constant approximation** of \( \lambda \).

---

In any case, we consider the corresponding distribution function:

$$
F : \mathbb{R} \rightarrow [0, 1], \quad
F(t) = 
\begin{cases}
1 - \exp\left( -\sum_{j=0}^{i-1} \lambda_j (t_{j+1} - t_j) + \lambda_i (t - t_i) \right), & t \geq 0 \\
0, & t < 0
\end{cases}
$$

where  
$$
i = i(t) := \max \{ i \mid t_i \le t \}.
$$

Then, \( F \) can be inverted efficiently.

> **How we can solve $\lambda_i (t - t_i)$ ?**

Given a uniform random variable $U$ on $(0,1)$, define:

$$
Z := -\log(1 - U)
$$

Determine $i$ such that:


$$
\Lambda(t_i) = \sum_{0 \leq j \leq i-1} \lambda_j (t_{j+1} - t_j) \leq Z < \sum_{0 \leq j \leq i} \lambda_j (t_{j+1} - t_j) = \Lambda(t_{i+1})
$$

and define:

$$
\tau = t_i + \frac{1}{\lambda_i} \left( Z - \sum_{0 \leq j \leq i-1} \lambda_j (t_{j+1} - t_j) \right)
$$

where:
- $ \Lambda (t_i) = \sum_{0 \leq j \leq i-1} \lambda_j (t_{j+1} - t_j)$ 

which is the solution of:

$$
-\log(1 - F(\tau)) = Z
$$

that is:

$$
F(\tau) = U
$$

Then, $(\tau)$ has a non-homogeneous exponential distribution with piecewise constant intensity $\tilde{\lambda}$, with:

$$
\tilde{\lambda}(t) = \lambda_i \quad \text{for} \quad t_i \leq t < t_{i+1}
$$
